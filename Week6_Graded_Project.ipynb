```python
# Import libraries
from transformers import pipeline, AutoTokenizer
import numpy as np
from gensim.downloader import load
from sklearn.metrics.pairwise import cosine_similarity
import warnings
warnings.filterwarnings('ignore')

print("✓ Libraries imported successfully")
```

```python
# Load the DistilGPT-2 model
print("Loading DistilGPT-2 model...")
generator = pipeline("text-generation", model="distilgpt2")
print("✓ Model loaded successfully\n")

# Prompt for generation
prompt = "AI is transforming industries by "
print(f"Prompt: '{prompt}'\n")
print("="*60)
print("GENERATED CONTINUATIONS:")
print("="*60)

# Generate three different continuations
outputs = generator(prompt, max_length=50, num_return_sequences=3, temperature=0.7)

for i, output in enumerate(outputs, 1):
    text = output['generated_text']
    continuation = text.replace(prompt, "").strip()
    print(f"\nContinuation {i}:")
    print(f"{prompt}{continuation}")
    print("-" * 60)
```

### Task 2: Tokenisation Demo

Tokenise a sentence and display tokens, token IDs, and sequence length.

```python
# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")

# Sentence to tokenise
sentence = "LLMs are powerful tools for natural language understanding."
print(f"Sentence: {sentence}\n")

# Tokenise the sentence
tokens = tokenizer.tokenize(sentence)
token_ids = tokenizer.encode(sentence)

# Display results
print("="*60)
print("TOKENIZATION RESULTS:")
print("="*60)
print(f"\nTokens: {tokens}")
print(f"\nToken IDs: {token_ids}")
print(f"\nSequence Length: {len(token_ids)} tokens")
print(f"\nToken-to-ID Mapping:")
print("-" * 40)
for token, token_id in zip(tokens, token_ids[:-1]):  # Exclude final token for clarity
    print(f"{token:20s} -> {token_id}")
```

Design and compare prompts for three different tasks.

```python
# ============================================================================
# SECTION B: PROMPT ENGINEERING CHALLENGE (COMPLETE FIXED CELL)
# ============================================================================
# Copy this ENTIRE cell into your Jupyter notebook Section B
# Replace everything from "print("="*70)" to the end of Task 3
# ============================================================================

from transformers import pipeline
import warnings
warnings.filterwarnings('ignore')

# CRITICAL FIX: Reload pipeline fresh for Section B
# This avoids inheriting defaults from Section A
print("\n" + "="*70)
print("Loading GPT-2 model (better than DistilGPT-2 for prompt engineering)")
print("="*70)
generator = pipeline("text-generation", model="gpt2")
print("✓ Model loaded\n")

# ============================================================================
# PROMPT ENGINEERING CHALLENGE: Comparing Task-Specific Prompts
# ============================================================================

print("="*70)
print("PROMPT ENGINEERING CHALLENGE: Comparing Task-Specific Prompts")
print("="*70)

# ============================================================================
# TASK 1: SUMMARISATION (≤ 30 words)
# ============================================================================
print("\n" + "="*70)
print("TASK 1: SUMMARISATION (≤ 30 words)")
print("="*70)

prompt1 = "Summarize in 30 words: Machine learning is a subset of artificial intelligence that enables systems to learn without being explicitly programmed."
print(f"\nPrompt: {prompt1}\n")

# FIX: Use ONLY max_new_tokens (not max_length), balanced temperature
output = generator(
    prompt1,
    max_new_tokens=35,  # Only this parameter - no max_length!
    num_return_sequences=1,
    temperature=0.7,
    do_sample=True,
    top_p=0.9
)
result = output[0]['generated_text'].replace(prompt1, "").strip()
print(f"Output: {result}")
print(f"Word count: {len(result.split())}")

# ============================================================================
# TASK 2: QUESTION & ANSWER
# ============================================================================
print("\n" + "="*70)
print("TASK 2: QUESTION & ANSWER")
print("="*70)

prompt2 = "Q: What is the capital of France?\nA: "
print(f"Prompt: {prompt2}\n")

# FIX: Very small max_new_tokens, low temperature, do_sample=False for factual output
output = generator(
    prompt2,
    max_new_tokens=10,  # VERY small - just want 1-2 words
    num_return_sequences=1,
    temperature=0,     # Deterministic
    do_sample=False    # No randomness - factual answers only
)
result = output[0]['generated_text'].replace(prompt2, "").strip()
print(f"Output: {result}")

# ============================================================================
# TASK 3: CREATIVE TEXT GENERATION (4-line poem on AI)
# ============================================================================
print("\n" + "="*70)
print("TASK 3: CREATIVE TEXT GENERATION (4-line poem on AI)")
print("="*70)

prompt3 = "Write a creative poem about artificial intelligence:\n"
print(f"Prompt: {prompt3}\n")

# FIX: Higher temperature for creativity, do_sample=True for diversity
output = generator(
    prompt3,
    max_new_tokens=60,  # Enough for poem lines
    num_return_sequences=1,
    temperature=0.7,   # Higher - more creative
    do_sample=True,     # Allow randomness
    top_p=0.95
)
result = output[0]['generated_text'].replace(prompt3, "").strip()
print(f"Output:\n{result}")
```

# Reflection on prompt engineering

**REFLECTION ON PROMPT ENGINEERING (100-150 words):**

Through this exercise, I observed several key patterns in how prompt rephrasing affects LLM outputs:

1. Task Clarity: More specific prompts (e.g., "Summarize in ≤30 words") guide the model toward
   constrained, focused outputs. Vague prompts result in longer, more rambling responses.

2. Prompt Structure: Adding action verbs ("Write", "Summarize", "Answer") helps the model
   interpret intent. The model responds differently to imperatives vs. declarative statements.

3. Context & Constraints: Specifying format (poem, 4 lines) or length limits significantly
   narrows the output space, making results more predictable and task-aligned.

4. Temperature Effects: Even with fixed prompts, temperature settings (0.7 here) introduce
   variability, showing the stochastic nature of generation.

Key Insight: Prompt engineering is fundamentally about reducing ambiguity and guiding the model's
probability distribution toward desired outputs. Small phrasing changes can dramatically shift results.


## SECTION C: Embeddings with Gensim

### Task 4: Word Embeddings

Use GloVe embeddings to explore word vectors and similarity.


```python
# Load GloVe embeddings
print("Loading GloVe embeddings (glove-wiki-gigaword-50)...")
wv = load('glove-wiki-gigaword-50')
print("✓ Embeddings loaded successfully\n")

# Select three words
words = ['king', 'queen', 'diamond']

print("="*70)
print("WORD EMBEDDINGS ANALYSIS")
print("="*70)

for word in words:
    if word in wv:
        print(f"\n{'='*70}")
        print(f"WORD: '{word.upper()}'")
        print(f"{'='*70}")

        # Get the embedding vector
        vector = wv[word]

        # Display first 10 values
        print(f"\nFirst 10 values of embedding vector:")
        print(vector[:10])
        print(f"\nVector dimension: {len(vector)}")

        # Get top 5 most similar words
        similar_words = wv.most_similar(word, topn=5)
        print(f"\nTop 5 most similar words and their similarity scores:")
        print("-"*50)
        for similar_word, score in similar_words:
            print(f"{similar_word:20s} -> Similarity: {score:.4f}")
    else:
        print(f"\nWord '{word}' not found in vocabulary")
```

### Task 5: Sentence-Level Embeddings

Compute average word vectors for sentences and create a similarity matrix.

```python
# Create five sentences
sentences = [
    "Artificial intelligence is transforming the world.",
    "Machine learning powers modern AI systems.",
    "Diamonds are precious gemstones.",
    "The queen wore a diamond crown on her head.",
    "AI and machine learning are related fields."
]

print("="*70)
print("SENTENCE-LEVEL EMBEDDINGS")
print("="*70)
print("\nSentences:")
for i, sent in enumerate(sentences, 1):
    print(f"{i}. {sent}")

# Function to get sentence embedding (average of word vectors)
def get_sentence_embedding(sentence, wv):
    words = sentence.lower().split()
    vectors = []
    for word in words:
        if word in wv:
            vectors.append(wv[word])

    if len(vectors) == 0:
        return np.zeros(wv.vector_size)

    return np.mean(vectors, axis=0)

# Compute sentence embeddings
sentence_embeddings = []
for sent in sentences:
    embedding = get_sentence_embedding(sent, wv)
    sentence_embeddings.append(embedding)

# Convert to numpy array
sentence_embeddings = np.array(sentence_embeddings)

print(f"\n✓ Sentence embeddings computed")
print(f"Embedding shape: {sentence_embeddings.shape} (5 sentences × 50 dimensions)")
```

```python
# Compute cosine similarity matrix
similarity_matrix = cosine_similarity(sentence_embeddings)

print("\n" + "="*70)
print("COSINE SIMILARITY MATRIX")
print("="*70)
print("\nSimilarity scores (0=dissimilar, 1=identical):")
print()

# Display matrix with labels
print("{:<5}".format(""), end="")
for i in range(1, 6):
    print("{:<12}".format(f"Sent{i}"), end="")
print()
print("-" * 70)

for i in range(5):
    print(f"Sent{i+1}", end="  ")
    for j in range(5):
        score = similarity_matrix[i][j]
        print("{:<12.4f}".format(score), end="")
    print()

# Identify most and least similar pairs
print("\n" + "-"*70)
print("INSIGHTS:")
print("-"*70)

# Find highest similarity (excluding diagonal)
max_sim = 0
max_pair = (0, 0)
for i in range(5):
    for j in range(i+1, 5):
        if similarity_matrix[i][j] > max_sim:
            max_sim = similarity_matrix[i][j]
            max_pair = (i, j)

print(f"\nMost similar pair (Sent{max_pair[0]+1}, Sent{max_pair[1]+1}): {max_sim:.4f}")
print(f"  Sent{max_pair[0]+1}: {sentences[max_pair[0]]}")
print(f"  Sent{max_pair[1]+1}: {sentences[max_pair[1]]}")

# Find lowest similarity
min_sim = 2
min_pair = (0, 1)
for i in range(5):
    for j in range(i+1, 5):
        if similarity_matrix[i][j] < min_sim:
            min_sim = similarity_matrix[i][j]
            min_pair = (i, j)

print(f"\nLeast similar pair (Sent{min_pair[0]+1}, Sent{min_pair[1]+1}): {min_sim:.4f}")
print(f"  Sent{min_pair[0]+1}: {sentences[min_pair[0]]}")
print(f"  Sent{min_pair[1]+1}: {sentences[min_pair[1]]}")
```
